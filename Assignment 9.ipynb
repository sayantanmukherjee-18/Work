```python

values = [3, 5, 2, 9, 12, 5, 23, 23]


# Minimax function using recursion
def minimax(depth, node_index, maximizing_player):

    # Base condition: leaf node
    if depth == 3:
        return values[node_index]

    # MAX player's turn
    if maximizing_player:
        return max(
            minimax(depth + 1, node_index * 2, False),
            minimax(depth + 1, node_index * 2 + 1, False)
        )

    # MIN player's turn
    else:
        return min(
            minimax(depth + 1, node_index * 2, True),
            minimax(depth + 1, node_index * 2 + 1, True)
        )


# Find optimal score for MAX
optimal_score = minimax(0, 0, True)

# Display result
print("Optimal score for MAX player:", optimal_score)
```

```python
# ============================================
# MINIMAX ALGORITHM
# Leaf Node Values:
# 3, 5, 2, 9, 12, 5, 23, 23
# ============================================

leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]


def minimax(depth, node_index, maximizing_player):
    # Base condition: leaf node
    if depth == 3:
        return leaf_values[node_index]

    if maximizing_player:
        best = float('-inf')

        for i in range(2):
            score = minimax(depth + 1, node_index * 2 + i, False)
            best = max(best, score)

        return best

    else:
        best = float('inf')

        for i in range(2):
            score = minimax(depth + 1, node_index * 2 + i, True)
            best = min(best, score)

        return best


# Calculate optimal score
best_score = minimax(0, 0, True)

print("Best possible score:", best_score)

```

```python

# ============================================
# ALPHA-BETA PRUNING ALGORITHM
# Leaf Node Values:
# 3, 5, 2, 9, 12, 5, 23, 23
# ============================================

leaf_values = [3, 5, 2, 9, 12, 5, 23, 23]


def alpha_beta(depth, node_index, maximizing_player, alpha, beta):

    # Base condition: leaf node
    if depth == 3:
        return leaf_values[node_index]

    if maximizing_player:
        best = float('-inf')

        for i in range(2):
            score = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                False,
                alpha,
                beta
            )

            best = max(best, score)
            alpha = max(alpha, best)

            # Beta cutoff
            if beta <= alpha:
                print("Pruned branch:", node_index * 2 + i + 1)
                break

        return best

    else:
        best = float('inf')

        for i in range(2):
            score = alpha_beta(
                depth + 1,
                node_index * 2 + i,
                True,
                alpha,
                beta
            )

            best = min(best, score)
            beta = min(beta, best)

            # Alpha cutoff
            if beta <= alpha:
                print("Pruned branch:", node_index * 2 + i + 1)
                break

        return best


# Initial values
alpha = float('-inf')
beta = float('inf')

# Find optimal score
best_score = alpha_beta(0, 0, True, alpha, beta)

print("\nBest possible score for MAX:", best_score)

```
